```python
datasets = {
    "A_raw": df_a,         # or sheet_a
    "A_aligned": None,
    "B_working": None
}
```

```python
pipeline = {
    "A_to_A_aligned": [
        schema_alignment,
        preprocess,
        rename_columns,
        drop_blanks,
        add_uid,
    ],
    "A_aligned_to_B": [
        remove_filters,
        preprocess,
        rename_columns,
        drop_blanks,
        automations,
    ],
    "Validation": [
        category_mapping,
        apply_rules,
        event_metadata,
        summary_stats,
    ]
}
```

```python
def run_pipeline(data, steps, kind="df"):
    for step in steps:
        data = step(data, kind=kind)
    return data
```

```python
datasets["A_aligned"] = run_pipeline(datasets["A_raw"], pipeline["A_to_A_aligned"])
datasets["B_working"] = run_pipeline(datasets["A_aligned"], pipeline["A_aligned_to_B"])
datasets["B_working"] = run_pipeline(datasets["B_working"], pipeline["Validation"])
```
